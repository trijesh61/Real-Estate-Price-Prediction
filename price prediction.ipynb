### Feature Engineering

```python
df2.head()
```

```python
df2["bhk"]=df2["size"].apply(lambda x: x.split(" ")[0])
df2.head()
```

```python
df2["total_sqft"].unique()
```

there are some unusual values are there

```python
def convert_sqft(x):
    f=x.split("-")
    if len(f)==2:
        return (float(f[0])+float(f[1]))/2
    try:
        return float(x)
    except:
        return None
```

```python
df3=df2.copy()
```

```python
df3.total_sqft = df3.total_sqft.apply(convert_sqft)
df3 = df3[df3.total_sqft.notnull()]
df3.head(2)
```

### Feature Engineering

```python
df4 = df3.copy()
df4['price_per_sqft'] = df3['price']*100000/df3['total_sqft']
df4.head()
```

```python
df4["price_per_sqft"].describe()
```

```python
df4["location"]=df4["location"].apply(lambda x: x.strip())
```
